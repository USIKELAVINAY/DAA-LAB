Implementation of making a change problem using dynamic progrmming

```python
import time


def min_coins(coins, amount):
    # dp[i] = minimum number of coins required to make amount i
    dp = [float('inf')] * (amount + 1)

    # 0 coins are required to make amount 0
    dp[0] = 0

    # Calculate minimum coins for every amount
    for i in range(1, amount + 1):
        for coin in coins:
            if coin <= i:
                dp[i] = min(dp[i], dp[i - coin] + 1)

    # If amount cannot be formed
    if dp[amount] == float('inf'):
        return -1

    return dp[amount]


# User input
n = int(input("Enter number of coins: "))

coins = list(map(int, input("Enter coin denominations: ").split()))

amount = int(input("Enter target amount: "))


# Measure execution time
start_time = time.perf_counter()

result = min_coins(coins, amount)

end_time = time.perf_counter()


# Display result
if result == -1:
    print("The amount cannot be formed.")
else:
    print("Minimum number of coins =", result)

# Execution time
execution_time = end_time - start_time
print("Execution time =", execution_time, "seconds")


# Complexity
print("Time Complexity: O(n × amount)")
print("Space Complexity: O(amount)")
```

Readme

This repository implements the classic Coin Change Problem using Dynamic Programming to find the minimum number of coins needed for a target amount. It optimizes the solution to \(O(N \times M)\) time complexity using a bottom-up tabulation approach.
