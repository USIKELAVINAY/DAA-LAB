MAXIMUM SORT AND MINIMUM SORT

```python
import time

# User input
n = int(input("Enter the number of elements: "))

arr = []
for i in range(n):
    value = int(input(f"Enter element {i + 1}: "))
    arr.append(value)

# Start execution timer
start_time = time.perf_counter()

# Find maximum
maximum = arr[0]
for i in range(1, n):
    if arr[i] > maximum:
        maximum = arr[i]

# Find minimum
minimum = arr[0]
for i in range(1, n):
    if arr[i] < minimum:
        minimum = arr[i]

# End execution timer
end_time = time.perf_counter()

execution_time = end_time - start_time

# Output
print("\nArray:", arr)
print("Maximum element:", maximum)
print("Minimum element:", minimum)
print("Execution time:", execution_time, "seconds")

print("\nTime Complexity: O(n)")
print("Space Complexity: O(n) for storing the input array")
```
