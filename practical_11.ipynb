Floyd–Warshall algorithm

```python

import time

INF = float('inf')

# User input
n = int(input("Enter the number of vertices: "))

print("Enter the adjacency matrix:")
print("Use INF for no direct edge and 0 for diagonal entries.")

dist = []

for i in range(n):
    row = input(f"Enter row {i + 1}: ").split()
    dist.append([
        INF if value.upper() == "INF" else float(value)
        for value in row
    ])

# Measure execution time of the algorithm
start_time = time.perf_counter()

# Floyd-Warshall algorithm
for k in range(n):
    for i in range(n):
        for j in range(n):
            if dist[i][k] != INF and dist[k][j] != INF:
                dist[i][j] = min(
                    dist[i][j],
                    dist[i][k] + dist[k][j]
                )

end_time = time.perf_counter()

# Display shortest distances
print("\nShortest distance matrix:")
for row in dist:
    print(" ".join(
        "INF" if value == INF else f"{value:g}"
        for value in row
    ))

# Check for a negative-weight cycle
if any(dist[i][i] < 0 for i in range(n)):
    print("Warning: Negative-weight cycle detected.")

# Execution time
execution_time = end_time - start_time
print(f"\nExecution time: {execution_time:.8f} seconds")

# Complexity
print("Time complexity: O(V^3)")
print("Space complexity: O(V^2)")
```

Readme:

The Floyd–Warshall algorithm finds the shortest paths between all pairs of vertices in a weighted graph using dynamic programming. This Python program accepts user input, displays the shortest-distance matrix, measures execution time, and demonstrates O(V³) time and O(V²) space complexity.
